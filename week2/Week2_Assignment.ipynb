**Schema**

```sql
-- Schema
DESCRIBE sample_superstore;
```

**Sample Data**

```sql
-- Sample Data
SELECT * FROM sample_superstore LIMIT 10;
```

```sql
-- Retrive Name and City of customers
Select `Customer Name`, City 
FROM sample_superstore
LIMIT 10;
```

**First and Last Name of all Customers with City**

```sql
-- Retrive only first_name,last_name and city of all customers
SELECT 
get(split(`Customer Name`,' '),0) AS First_Name,
get(split(`Customer Name`,' '),1) AS Last_Name, 
City 
FROM sample_superstore
LIMIT 10;
```

**Unique Categories**

```sql
-- Unique Categories  available in superstore table
SELECT DISTINCT Category
From sample_superstore;
```

**Number of Unique Order IDs**

```sql
SELECT 
COUNT(*) AS Total_rows,
COUNT(DISTINCT `Order ID`) AS Unique_OrderIDs
FROM sample_superstore;
```

**WHERE filters**

```sql
-- Filter the data for South region
SELECT * FROM sample_superstore
WHERE Region='South'
LIMIT 10;
```

```sql
-- Filter the data for Technology category
SELECT * FROM sample_superstore
WHERE Category='Technology'
LIMIT 10;
```

```sql
--  Orders between 2016 and 2017
SELECT * FROM sample_superstore
WHERE DATE(`Order Date`)
BETWEEN '2016-01-01' AND '2017-01-01'
LIMIT 10;
```

```sql
-- Sales >1000
SELECT * FROM sample_superstore
WHERE `Sales`>1500
LIMIT 10;
```

```sql
-- Rows where region is South, category is Furniture and sales >=500
SELECT * FROM sample_superstore
Where Region='South'
And Category='Furniture'
And Sales>=500
LIMIT 10;
```

```sql
--Loss Making Orders
SELECT *
FROM sample_superstore
WHERE Profit<0
LIMIT 10;
```

**GROUP BY**

```sql
-- Total Sales by Region
SELECT Region,
       SUM(Sales) AS Total_Sales
FROM sample_superstore
GROUP BY Region;
```

```sql
-- Total Profit by Region
SELECT Region,
       SUM(Profit) AS Total_Profit
FROM sample_superstore
GROUP BY Region;
```

```sql
-- Average Sales by Category
SELECT Category,
       AVG(Sales) AS Avg_Sales
FROM sample_superstore
GROUP BY Category;
```

```sql
-- Total Quantity Sold by Category
SELECT Category,
       SUM(Quantity) AS Total_Quantity
FROM sample_superstore
GROUP BY Category;
```

```sql
-- Sales and Profit by Category
SELECT Category,
       SUM(Sales) AS Total_Sales,
       SUM(Profit) AS Total_Profit
FROM sample_superstore
GROUP BY Category;
```

**Sorting and Limiting Results**

```sql
-- Top 10 Highest Sales Orders
SELECT *
FROM sample_superstore
ORDER BY Sales DESC
LIMIT 10;
```

```sql
-- Top 5 Profitable Sub-Categories by Category
SELECT `Sub-Category`,Category,
       SUM(Profit) AS Profit
FROM sample_superstore
GROUP BY `Sub-Category`,Category
ORDER BY Profit DESC
LIMIT 5;
```

```sql
-- Top 10 Products by Sales Revenue
SELECT `Product Name`,
       SUM(Sales) AS Total_Sales
FROM sample_superstore
GROUP BY `Product Name`
ORDER BY Total_Sales DESC
LIMIT 10;
```

```sql
-- Top Categories by Profit
SELECT Category,
       SUM(Profit) AS Total_Profit
FROM sample_superstore
GROUP BY Category
ORDER BY Total_Profit DESC;
```

**Business Use Case**

```sql
-- Monthly Sales Trend Analysis
SELECT YEAR(`Order Date`) AS Year,
       MONTH(`Order Date`) AS Month,
       SUM(Sales) AS Monthly_Sales
FROM sample_superstore
GROUP BY YEAR(`Order Date`),
         MONTH(`Order Date`)
ORDER BY Year, Month;
```

```sql
-- Orders in Every Year
SELECT YEAR(`Order Date`) AS Year,
       COUNT(*) AS Orders
FROM sample_superstore
GROUP BY YEAR(`Order Date`)
ORDER BY Year;
```

```sql
-- Top Customers by Sales
SELECT `Customer Name`,
       SUM(Sales) AS Total_Sales
FROM sample_superstore
GROUP BY `Customer Name`
ORDER BY Total_Sales DESC
LIMIT 10;
```

```sql
-- Duplicate Order IDs
SELECT `Order ID`,
       COUNT(*) AS Record_Count
FROM sample_superstore
GROUP BY `Order ID`
HAVING COUNT(*) > 1
LIMIT 20;
```

**Data Validation and Quality Checks**

```sql
-- Total Rows
SELECT COUNT(*) AS Total_Rows
FROM sample_superstore;
```

```sql
-- Invalid Shipping Dates
SELECT *
FROM sample_superstore
WHERE `Ship Date` < `Order Date`;
```

```sql
-- Identify Primary Key
SELECT 
COUNT(*) AS Total_rows,
COUNT(DISTINCT `Row ID`) AS Unique_RowIDs
FROM sample_superstore;
```

```sql
-- Check for Null Values
SELECT
SUM(CASE WHEN `Order ID` IS NULL THEN 1 ELSE 0 END) AS Null_OrderID,
SUM(CASE WHEN `Customer Name` IS NULL THEN 1 ELSE 0 END) AS Null_Customer,
SUM(CASE WHEN Sales IS NULL THEN 1 ELSE 0 END) AS Null_Sales,
SUM(CASE WHEN Profit IS NULL THEN 1 ELSE 0 END) AS Null_Profit
FROM sample_superstore;
```

```sql
-- Orders with Zero Sales
SELECT *
FROM sample_superstore
WHERE Sales = 0;
```
