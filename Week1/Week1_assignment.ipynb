```python
#count of missing values in each column before handling missing values
df.isnull().sum()
```

```python
#filling discount column missing values with median
df["discount"]=df["discount"].fillna(df["discount"].median())
df
```

```python
#filling seller_name column with unknown seller
df["seller_name"]=df["seller_name"].fillna("unknown seller")
df
```

```python
#filling missing values
df["what_customers_said"]=df["what_customers_said"].fillna("no reviews")
df["videos"]=df["videos"].fillna("no video")
df["seller_information"]=df["seller_information"].fillna("no available")
df["variations"]=df["variations"].fillna("no variations")
df
```

```python
#count after handling missing values
df.isnull().sum()
```

```python
#select columns
df[["title","category","ratings_count"]].head()
```

```python
#filter rows where discount > 50
df[df["discount"]>50]
```

```python
#filter rows print category,rating where ratings>4
df[df["rating"]>4][["category","rating"]]
```

```python
#filter rows where category=bedsheets with sizes
df[df["category"]=="bedsheets"][["sizes","category"]]
```

```python
#removing duplicates as there are no duplicates sum is 0 before and after
print("Sum of duplicates before removing:",df.duplicated().sum())
df=df.drop_duplicates()
print("Sum of duplicates after removing:",df.duplicated().sum())
```

```python
#converting final_price column from string to float
df["final_price"]=pd.to_numeric(df["final_price"].str.replace("₹","").str.replace(",","").str.replace('"',""))
df
```

```python
#creating a new column savings dervied from initial and final price
df["savings"]=df["initial_price"]-df["final_price"]
df
```

```python
#displaying savings of last 10 rows
df[["initial_price","final_price","savings"]].tail(10)
```

```python
#savings of first 20 rows
df[["initial_price","final_price","savings"]].head(20)
```
